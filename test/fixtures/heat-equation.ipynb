# The Heat Equation

We solve the one-dimensional heat equation

$$\frac{\partial u}{\partial t} = D \frac{\partial^2 u}{\partial x^2}$$

on $0 \le x \le L$ with $u(0, t) = u(L, t) = 0$, using explicit finite
differences on a grid $x_j = j \Delta x$, $t_n = n \Delta t$.

## Finite Differences

With $u_j^n \approx u(x_j, t_n)$, the forward difference in time and the
central difference in space give

$$u_j^{n+1} = u_j^n + \sigma \left(u_{j+1}^n - 2 u_j^n + u_{j-1}^n\right),
\qquad \sigma = \frac{D \Delta t}{\Delta x^2}.$$

The scheme is stable for $\sigma \le \frac{1}{2}$.

```python
import numpy as np
from matplotlib import pyplot as plt
%matplotlib inline
```

### Grid and Initial Condition

```python
L, D = 1.0, 0.1
J, N = 50, 400
dx = L / J
dt = 0.4 * dx**2 / D
sigma = D * dt / dx**2
print("sigma =", sigma)
```

```python
x = np.linspace(0, L, J + 1)
u = np.sin(np.pi * x)
u[[0, -1]]
```

### Time Stepping

```python
def step(u, sigma):
    v = u.copy()
    v[1:-1] = u[1:-1] + sigma * (u[2:] - 2 * u[1:-1] + u[:-2])
    return v

for n in range(N):
    u = step(u, sigma)
```

## Results

The exact solution is $u(x, t) = e^{-D \pi^2 t} \sin(\pi x)$.

```python
plt.plot(x, u)
```

```python
exact = np.exp(-D * np.pi**2 * N * dt) * np.sin(np.pi * x)
plt.plot(x, u - exact)
```

```python
plt.semilogy(x[1:-1], abs(u - exact)[1:-1])
```

The error is of order $\Delta t + \Delta x^2$.
